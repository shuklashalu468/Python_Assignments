```python
# 1) Use plt.subplot() to create a figure with 2 rows and 1 column of plots showing your daily step count and water intake (in litres) 
# for the last 7 days. Plot both as line charts, one in each subplot.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

steps = [6500, 7200, 8100, 9000, 7500, 6800, 8500]

water = [2, 2.5, 2, 3, 2.5, 3, 2.8]

plt.figure(figsize=(8, 7))

plt.subplot(2, 1, 1)
plt.plot(days, steps, marker="o")
plt.xlabel("Days")
plt.ylabel("Steps")
plt.title("Daily Step Count")

plt.subplot(2, 1, 2)
plt.plot(days, water, marker="o")
plt.xlabel("Days")
plt.ylabel("Water Intake (Litres)")
plt.title("Daily Water Intake")

plt.tight_layout()
plt.show()
```

```python
# 2) Create a 2x2 multi-panel figure using plt.subplots(). In each subplot, display the number of Instagram posts you made in each 
# quarter (Q1, Q2, Q3, Q4) for the past year, using different chart types: line, bar, scatter, and pie.

quarters = ["Q1", "Q2", "Q3", "Q4"]
posts = [25, 40, 30, 50]

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

ax[0, 0].plot(quarters, posts, marker="o")
ax[0, 0].set_title("Line Chart")
ax[0, 0].set_xlabel("Quarter")
ax[0, 0].set_ylabel("Number of Posts")

ax[0, 1].bar(quarters, posts)
ax[0, 1].set_title("Bar Chart")
ax[0, 1].set_xlabel("Quarter")
ax[0, 1].set_ylabel("Number of Posts")

ax[1, 0].scatter(quarters, posts)
ax[1, 0].set_title("Scatter Plot")
ax[1, 0].set_xlabel("Quarter")
ax[1, 0].set_ylabel("Number of Posts")

ax[1, 1].pie(posts, labels=quarters, autopct="%1.1f%%")
ax[1, 1].set_title("Pie Chart")

plt.tight_layout()
plt.show()
```

```python
# 3) Make a grid of 1 row and 3 columns using plt.subplots(), and plot the number of food orders you placed on Zomato, Swiggy, and 
# Domino's over the last 6 months, one app per subplot. Share the y-axis among all subplots so you can easily compare order counts 
# across apps.<br><br><em><strong>Hint:</strong> Use the 'sharey' parameter in plt.subplots().</em>

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

zomato = [12, 15, 10, 18, 20, 16]
swiggy = [10, 14, 12, 15, 17, 19]
dominos = [5, 8, 6, 10, 12, 9]

fig, ax = plt.subplots(1, 3, figsize=(15, 5), sharey=True)

ax[0].plot(months, zomato, marker="o")
ax[0].set_title("Zomato Orders")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Number of Orders")

ax[1].plot(months, swiggy, marker="o")
ax[1].set_title("Swiggy Orders")
ax[1].set_xlabel("Month")

ax[2].plot(months, dominos, marker="o")
ax[2].set_title("Domino's Orders")
ax[2].set_xlabel("Month")

plt.tight_layout()
plt.show()
```
