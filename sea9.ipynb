```python
# 1) Use Seaborn's regplot to visualize the relationship between the number of likes and number of comments on a set of Instagram posts 
# (create a small DataFrame with at least 10 sample data points).

import pandas as pd

data = {
    "likes": [500, 800, 1200, 1500, 2000, 2500, 3000, 3500, 4200, 5000],
    "comments": [20, 30, 45, 55, 70, 85, 100, 120, 140, 160]
}

df = pd.DataFrame(data)

print(df)

import seaborn as sns
import matplotlib.pyplot as plt

sns.regplot(
    data=df,
    x="likes",
    y="comments"
)

plt.xlabel("Number of Likes")
plt.ylabel("Number of Comments")
plt.title("Instagram Likes vs Comments")

plt.show()
```

```python
# 2) Create a lmplot in Seaborn to show the regression line between food order price and delivery time using a sample Zomato-style 
# dataset (minimum 15 rows with random values for price and delivery time).

data = {
    "order_price": [
        150, 220, 300, 180, 450,
        500, 350, 600, 250, 700,
        400, 550, 800, 650, 900
    ],
    
    "delivery_time": [
        20, 25, 28, 22, 35,
        38, 30, 42, 26, 45,
        33, 40, 50, 44, 55
    ]
}

df = pd.DataFrame(data)

sns.lmplot(
    data=df,
    x="order_price",
    y="delivery_time"
)

plt.xlabel("Food Order Price (₹)")
plt.ylabel("Delivery Time (minutes)")
plt.title("Food Order Price vs Delivery Time")

plt.show()

```
