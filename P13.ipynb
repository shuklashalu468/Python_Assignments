```python
# 2:
mean_runs = df["runs"].mean()
median_runs = df["runs"].median()
mode_runs = df["runs"].mode()
min_runs = df["runs"].min()
max_runs = df["runs"].max()

print("Mean:", mean_runs)
print("Median:", median_runs)
print("Mode:", mode_runs.tolist())
print("Minimum:", min_runs)
print("Maximum:", max_runs)
```
