```python
import numpy as np

prices = np.array([199, 299, 399, 499, 599])

print("First price:", prices[0])
print("Last price:", prices[-1])
```

```python
scores = np.array([
    [45, 60, 72, 80, 55],
    [50, 65, 70, 85, 60],
    [40, 55, 75, 90, 65]
])

# Matches 2 to 4 → index 1 to 3
result = scores[:, 1:4]

print(result)
```

```python
ratings = np.array([4.5, 3.8, 4.2, 2.9, 5.0, 3.5])

print(ratings[-3:])
```

```python
numbers = np.arange(1, 21)

result = numbers[1::3]

print(result)
```

```python
prices = np.array([250, 650, 450, 800, 550, 300, 1000])

result = prices[prices > 500]

print(result)
```

```python
scores = np.array([210, 180, 195, 220, 205, 175])

result = np.where(scores > 200)

print(result)
```
