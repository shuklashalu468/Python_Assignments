```python
# Dictionary
playlist_prices = {
    "Top Hits": 199,
    "Chill Vibes": 149,
    "Workout Mix": 249,
    "Party Songs": 299,
    "Classical Relax": 99
}

# Function to update the playlist price
def update_playlist_price(playlist, new_price):
    playlist_prices[playlist] = new_price

# Update the price of one playlist
update_playlist_price("Top Hits", 250)

# Print the updated dictionary
print("Updated Playlist Prices:")
print(playlist_prices)
```

```python
del playlist_prices["Party Songs"]

# Print the updated dictionary
print("Dictionary after deletion:")
print(playlist_prices)
```

```python
# Restaurants ordered from Zomato
set1 = {"Domino's", "KFC", "McDonald's", "Burger King"}

# Restaurants ordered from Swiggy
set2 = {"KFC", "Subway", "Burger King", "Pizza Hut"}

# Find the union of the two sets
all_restaurants = set1.union(set2)

# Find the intersection of the two sets
common_restaurants = set1.intersection(set2)

# Print the results
print("Union:", all_restaurants)
print("Intersection:", common_restaurants)
```
