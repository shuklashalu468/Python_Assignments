```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 2) Using Seaborn's lineplot(), plot the daily temperature trend for one week using the following data: 
# days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'], temps = [32, 34, 33, 35, 36, 37, 34]. 
# Label the axes and title the chart 'Weekly Temperature Trend'.

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps)

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend")

plt.show()
```

```python
# 3) Recreate the above line plot using Seaborn's axes-level and figure-level functions separately. 
# Briefly note one difference you observe in how these functions work.
# <br><br><em><strong>Hint:</strong> Use sns.lineplot() for axes-level and sns.relplot() for figure-level.</em>

sns.lineplot(x=days, y=temps)

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend")

plt.show()
# sns.lineplot() is an axes-level function, so it draws the plot on a specific Axes.

sns.relplot(x=days, y=temps, kind="line")

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend")

plt.show()

# sns.relplot() is a figure-level function, so it creates and manages the overall figure and is useful when you want multiple plots using options like row or col.
```

```python
# 4) Apply each of Seaborn's built-in themes (darkgrid, whitegrid, ticks) to your temperature trend plot and save three separate images.
# Compare how the grid and background change with each theme.<br><br><em><strong>Hint:</strong> 
# Use sns.set_theme(style='theme_name') before plotting.</em>

# Darkgrid theme

sns.set_theme(style="darkgrid")

sns.lineplot(x=days, y=temps)

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend - Darkgrid")

plt.savefig("temperature_darkgrid.png")
plt.show()

# Whitegrid theme

sns.set_theme(style="whitegrid")

sns.lineplot(x=days, y=temps)

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend - Whitegrid")

plt.savefig("temperature_whitegrid.png")
plt.show()

# Ticks theme

sns.set_theme(style="ticks")

sns.lineplot(x=days, y=temps)

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("Weekly Temperature Trend - Ticks")

plt.savefig("temperature_ticks.png")
plt.show()
```
