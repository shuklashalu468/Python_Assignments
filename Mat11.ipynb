```python
# 1) Create a simple bar chart showing the number of orders for 5 different Zomato restaurants and save the chart as a PNG file using 
# plt.savefig() with a DPI of 150.

import matplotlib.pyplot as plt

restaurants = ["Restaurant A", "Restaurant B", "Restaurant C", "Restaurant D", "Restaurant E"]
orders = [120, 150, 90, 180, 130]

plt.bar(restaurants, orders)

plt.xlabel("Restaurants")
plt.ylabel("Number of Orders")
plt.title("Zomato Restaurant Orders")

plt.savefig("zomato_orders.png", dpi=150)

plt.show()
```

```python
# 2) Generate a line plot that displays your daily step count for one week (7 days) and export it as an SVG file using plt.savefig()
# .<br><br><em><strong>Hint:</strong> Change the file extension in the filename argument to save as SVG

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [6500, 7200, 8100, 9000, 7500, 6800, 8500]

plt.plot(days, steps, marker="o")

plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.title("Daily Step Count for One Week")

plt.savefig("weekly_steps.svg")

plt.show()
```

```python
# 3) Build a pie chart showing the percentage of time you spend on 4 different apps (Instagram, YouTube, WhatsApp, Spotify) in a day 
# and save the chart as a PDF using plt.savefig().

apps = ["Instagram", "YouTube", "WhatsApp", "Spotify"]
time_spent = [3, 2.5, 2, 1.5]

plt.pie(time_spent, labels=apps, autopct="%1.1f%%")

plt.title("Daily Time Spent on Different Apps")

plt.savefig("daily_app_usage.pdf")

plt.show()
```
