```python
# 3) Given an array of daily step counts from a fitness tracker app for one month, plot two histograms using plt.hist(): one showing 
# frequency (default), and one showing density=True. Compare the y-axis labels and explain the difference.

steps = [6500, 7200, 8100, 9000, 7500, 6800, 10200, 9500, 8700, 7300,
         8200, 7800, 6900, 10500, 9100, 8500, 7600, 9400, 8800, 7900,
         6700, 7300, 9700, 8300, 8900, 7100, 9200, 8000, 8600, 7400]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.hist(steps, bins=6)
plt.xlabel("Daily Steps")
plt.ylabel("Frequency")
plt.title("Step Counts - Frequency")

plt.subplot(1, 2, 2)
plt.hist(steps, bins=6, density=True)
plt.xlabel("Daily Steps")
plt.ylabel("Density")
plt.title("Step Counts - Density")

plt.tight_layout()
plt.show()
```

```python
# 4) Simulate the distribution of song durations (in seconds) from your Spotify playlist by generating 50 random numbers between 120 
# and 300. Plot a histogram using plt.hist() and set the color to 'purple' and add the title 'Spotify Song Duration Distribution'.

import numpy as np

song_durations = np.random.randint(120, 301, 50)

plt.hist(song_durations, color="purple")

plt.xlabel("Song Duration (Seconds)")
plt.ylabel("Number of Songs")
plt.title("Spotify Song Duration Distribution")

plt.show()
```
