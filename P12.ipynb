```python
# 1:

import pandas as pd

data = {
    "team": ["MI", "CSK", "RCB", "MI", "CSK", "RCB", "KKR", "MI"],
    "runs": [180, 160, 190, 200, 170, 150, 175, 210]
}

df = pd.DataFrame(data)

df.to_csv("ipl_matches.csv", index=False)

print("IPL CSV file created successfully!")
```

```python
df = pd.read_csv("ipl_matches.csv")

print(df)
```

```python
total_runs = df.groupby("team")["runs"].sum()

total_runs = pd.DataFrame(total_runs).reset_index()

print(total_runs)
```

```python
# 2:

data = {
    "city": ["Delhi", "Delhi", "Delhi", "Mumbai", "Mumbai", "Mumbai",
             "Pune", "Pune"],
    "restaurant": ["Dominos", "Dominos", "Burger King", "Dominos",
                   "Burger King", "Burger King", "Dominos", "Burger King"],
    "order_amount": [300, 500, 400, 350, 450, 550, 250, 350]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df.groupby(["city", "restaurant"]).agg(average_order_amount=("order_amount", "mean")).reset_index()

print(result)
```

```python
# 3:

data = {
    "category": ["Electronics", "Electronics", "Fashion", "Fashion",
                 "Electronics", "Fashion", "Electronics", "Fashion"],
    "region": ["North", "North", "North", "South",
               "South", "South", "North", "North"],
    "sales": [50000, 30000, 20000, 15000, 40000, 25000, 35000, 18000],
    "units_sold": [10, 6, 8, 5, 9, 7, 7, 6]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df.groupby(["category", "region"]).agg(
    total_sales=("sales", "sum"),
    average_units_sold=("units_sold", "mean")
).reset_index()

print(result)
```

```python
# 4:

data = {
    "region": ["North", "North", "South", "South", "East", "East"],
    "revenue": [50000, 70000, 40000, 60000, 30000, 45000]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df.groupby("region").agg({
    "revenue": ["sum", "max", "min"]
})

print(result)
```

```python
# 5:

data = {
    "artist": ["Taylor Swift", "Ed Sheeran", "Dua Lipa",
               "Arijit Singh", "The Weeknd", "Drake", "Ed Sheeran"],
    "genre": ["Pop", "Pop", "Pop", "Bollywood", "R&B", "Hip-Hop", "Pop"],
    "streams": [5000000, 4000000, 3000000, 2500000, 3500000, 4500000, 2000000]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df.groupby("genre").agg(
    total_streams=("streams", "sum"),
    average_streams=("streams", "mean")
)

result = result.sort_values("total_streams", ascending=False)

print(result)
```
