```python
# 3) Import Seaborn and use sns.histplot() to visualize the distribution of delivery times (in minutes) from a sample 
# Swiggy order dataset with at least 30 values.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "delivery_time": [
        20, 25, 30, 35, 40, 28, 32, 45, 22, 27,
        30, 35, 38, 42, 25, 29, 33, 37, 40, 24,
        26, 31, 34, 36, 43, 21, 23, 28, 39, 44,
        32, 30, 27, 35, 41
    ]
}

df = pd.DataFrame(data)

sns.histplot(data=df, x="delivery_time")

plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")
plt.title("Distribution of Swiggy Delivery Times")

plt.show()
```

```python
# 2) Download or create a list of daily step counts (e.g., from a fitness app) for one month and plot the 
# distribution using sns.kdeplot(), showing the probability density curve.

steps = [
    5200, 6800, 7500, 8200, 6100,
    9000, 7200, 5600, 8800, 9500,
    7000, 6400, 8100, 7600, 10300,
    5900, 6800, 8700, 9200, 7400,
    6500, 7800, 8500, 9600, 7200,
    5500, 6900, 8000, 8900, 9800
]

df = pd.DataFrame({"steps": steps})

print(df)

sns.kdeplot(data=df, x="steps")

plt.xlabel("Daily Steps")
plt.ylabel("Probability Density")
plt.title("Distribution of Daily Step Counts")

plt.show()
```

```python
# 3) Combine sns.histplot() and sns.kdeplot() to visualize the distribution of monthly spending amounts (in rupees) on 
# Zomato orders, using the same dataset for both plots on a single figure.<br><br><em><strong>Hint:</strong> Use the 
# same data and overlay the KDE curve on the histogram by calling both functions on the same axes.</em>

data = {
    "monthly_spending": [
        450, 800, 1200, 650, 1500,
        900, 1100, 700, 1800, 1300,
        950, 600, 1600, 1250, 850,
        1400, 1000, 750, 2000, 1150,
        550, 1350, 900, 1700, 1050,
        1250, 800, 1450, 1900, 700
    ]
}

df = pd.DataFrame(data)

print(df)

sns.histplot(
    data=df,
    x="monthly_spending",
    bins=8,
    kde=False
)

sns.kdeplot(
    data=df,
    x="monthly_spending"
)

plt.xlabel("Monthly Spending (₹)")
plt.ylabel("Density / Number of Orders")
plt.title("Distribution of Monthly Zomato Spending")

plt.show()
```

```python
# 4) Given a dataset of IPL match scores, use sns.histplot() to plot the run distribution. Change the number of bins and 
# observe how the plot changes. Write 2-3 lines about what you notice.<br><br><em><strong>Hint:</strong> Try bins=5,
# bins=10, and bins=20.</em>

data = {
    "runs": [
        120, 145, 160, 175, 190, 135, 155, 180, 200, 210,
        125, 150, 165, 170, 185, 195, 220, 230, 140, 158,
        172, 188, 205, 215, 132, 148, 162, 178, 192, 225
    ]
}

df = pd.DataFrame(data)

print(df)

# 5 bins

sns.histplot(data=df, x="runs", bins=5)

plt.xlabel("Runs")
plt.ylabel("Number of Matches")
plt.title("IPL Run Distribution - 5 Bins")
plt.show()

# 10 bins

sns.histplot(data=df, x="runs", bins=10)

plt.xlabel("Runs")
plt.ylabel("Number of Matches")
plt.title("IPL Run Distribution - 10 Bins")
plt.show()

# 20 bins

sns.histplot(data=df, x="runs", bins=20)

plt.xlabel("Runs")
plt.ylabel("Number of Matches")
plt.title("IPL Run Distribution - 20 Bins")
plt.show()

# With 5 bins, the distribution looks more simple and gives a broad overview of the scores.
# With 10 bins, more details about the run distribution become visible.
# With 20 bins, the graph becomes more detailed but can look crowded.
```
