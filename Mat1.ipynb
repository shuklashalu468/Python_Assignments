```python
# 1) Use plt.plot() to create a line chart showing the number of daily steps you walked over the last 7 days. 
# Set the line color to green and linewidth to 2.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [5000, 6500, 7200, 6000, 8000, 9500, 7500]

plt.plot(days, steps, color="green", linewidth=2)

plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.title("Daily Steps Over 7 Days")

plt.show()
```

```python
# 2) Plot a line chart of your favorite music app's monthly active users for 6 months using plt.plot(). Add circle markers at each data 
# point and set the marker color to red.<br><br><em><strong>Hint:</strong> Use the marker and markerfacecolor parameters in plt.plot().

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
users = [120, 150, 180, 170, 220, 250]

plt.plot(
    months,
    users,
    marker="o",
    markerfacecolor="red"
)

plt.xlabel("Month")
plt.ylabel("Monthly Active Users (in millions)")
plt.title("Monthly Active Users of My Favorite Music App")

plt.show()
```

```python
# 3) Create a line chart comparing the number of orders you placed on Zomato and Swiggy over the last 5 months. Use different line 
# styles and colors for each app, and add a legend to distinguish them.

months = ["Jan", "Feb", "Mar", "Apr", "May"]

zomato_orders = [12, 15, 18, 14, 20]
swiggy_orders = [10, 13, 16, 19, 17]

plt.plot(
    months,
    zomato_orders,
    color="red",
    linestyle="-",
    marker="o",
    label="Zomato"
)

plt.plot(
    months,
    swiggy_orders,
    color="orange",
    linestyle="--",
    marker="o",
    label="Swiggy"
)

plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.title("Zomato vs Swiggy Orders")

plt.legend()

plt.show()
```
