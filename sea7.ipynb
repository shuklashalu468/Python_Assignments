```python
# 1) Create a Pandas DataFrame with at least 5 columns representing different features of food delivery orders (e.g., delivery_time, 
# order_value, tip_amount, rating, delivery_distance) and 20 rows of random data, then calculate and display the correlation matrix
# using the .corr() function.

import pandas as pd
import numpy as np

np.random.seed(42)

data = {
    "delivery_time": np.random.randint(15, 60, 20),
    "order_value": np.random.randint(200, 1500, 20),
    "tip_amount": np.random.randint(10, 200, 20),
    "rating": np.round(np.random.uniform(2.5, 5.0, 20), 1),
    "delivery_distance": np.round(np.random.uniform(1, 15, 20), 2)
}

df = pd.DataFrame(data)

print(df)

correlation_matrix = df.corr()

print(correlation_matrix)
```

```python
# 2) Using Seaborn, plot a heatmap of the correlation matrix you generated for the food delivery data, and add annotations to show the 
# correlation values inside each cell.

import seaborn as sns

sns.heatmap(
    correlation_matrix,
    annot=True
)

plt.title("Correlation Heatmap of Food Delivery Data")
plt.show()
```

```python
# 3) Change the color palette of your Seaborn heatmap to a diverging palette (such as 'coolwarm' or 'RdYlGn') to better visualize 
# positive and negative correlations, and explain which two features in your data have the strongest negative correlation.
# <br><br><em><strong>Hint:</strong> Use the cmap parameter in sns.heatmap().</em>

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Food Delivery Data")
plt.show()

# The strongest negative correlation is between tip_amount and rating, with a correlation of approximately -0.477. This indicates a
# moderate negative relationship, meaning the two variables tend to move in opposite directions in this sample.
```
