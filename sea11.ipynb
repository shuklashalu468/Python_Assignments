```python
# 1) Create a Seaborn barplot showing the number of orders per food category (e.g., Pizza, Burger, Biryani, Momos, Ice Cream) and 
# save the chart as a PNG file with 300 DPI.

import pandas as pd

data = {
    "food_category": [
        "Pizza", "Burger", "Biryani", "Momos", "Ice Cream"
    ],
    "orders": [
        120, 95, 110, 80, 70
    ]
}

df = pd.DataFrame(data)

import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(
    data=df,
    x="food_category",
    y="orders"
)

plt.xlabel("Food Category")
plt.ylabel("Number of Orders")
plt.title("Number of Orders by Food Category")

plt.savefig("food_category_orders.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
# 2) Generate a Seaborn line plot visualizing daily step counts for one week, then export the chart as an SVG file suitable
#     for use in a PowerPoint presentation.<br><br><em><strong>Hint:</strong> Use plt.savefig('filename.svg', format='svg'). </em>

data = {
    "day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "steps": [5200, 6800, 7500, 8200, 6100, 9000, 7800]
}

df = pd.DataFrame(data)

sns.lineplot(
    data=df,
    x="day",
    y="steps",
    marker="o"
)

plt.xlabel("Day")
plt.ylabel("Daily Steps")
plt.title("Daily Step Counts - One Week")

plt.savefig("daily_steps.svg", format="svg", bbox_inches="tight")

plt.show()
```

```python
# 3) Customize a Seaborn chart using the 'darkgrid' style and a color palette of your choice, then save it as a high-resolution PNG 
# image with 1200 DPI.

data = {
    "category": ["Pizza", "Burger", "Biryani", "Momos", "Ice Cream"],
    "orders": [120, 95, 110, 80, 70]
}

df = pd.DataFrame(data)

sns.set_theme(
    style="darkgrid",
    palette="deep"
)

sns.barplot(
    data=df,
    x="category",
    y="orders",
    hue="category",
    legend=False
)

plt.xlabel("Food Category")
plt.ylabel("Number of Orders")
plt.title("Food Orders by Category")

plt.savefig(
    "food_orders_high_resolution.png",
    dpi=1200,
    bbox_inches="tight"
)

plt.show()
```

```python
# 4) Create a template function called apply_presentation_theme() that applies a consistent Seaborn style, font, and color palette
# to any chart, then use it to style and export a pie chart showing yourtop 5 most-used apps.

def apply_presentation_theme():
    sns.set_theme(
        style="whitegrid",
        context="talk",
        palette="deep",
        font_scale=1.1
    )

data = {
    "app": ["Instagram", "YouTube", "WhatsApp", "Spotify", "Netflix"],
    "usage_hours": [18, 15, 12, 8, 6]
}

df = pd.DataFrame(data)

apply_presentation_theme()

plt.figure(figsize=(8, 8))

plt.pie(
    df["usage_hours"],
    labels=df["app"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Most-Used Apps")

plt.savefig(
    "top_5_most_used_apps.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
