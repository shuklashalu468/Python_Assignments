```python
# 1:
import pandas as pd

data = {
    "username": ["CoolDude", "foodieQueen", "TravelGirl", "FashionSTAR", "TechBuddy"]
}

df = pd.DataFrame(data)

print("Original usernames:")
print(df)
```

```python
df["username"] = df["username"].str.lower()

print("Lowercase usernames:")
print(df)
```

```python
# 2:
songs = pd.Series([
    "Shape of You Remix",
    "Blinding Lights Remix",
    "Perfect",
    "Levitating Remix",
    "Believer Remix"
])

print(songs)
```

```python
cleaned_songs = songs.str.replace("Remix", "", regex=False)

print(cleaned_songs)
```

```python
# 3:
data = {
    "review": [
        "The food was good and delivery was fast",
        "Delivery was very late",
        "Amazing food and great taste",
        "The delivery boy was polite",
        "Very expensive restaurant"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# 4:
data = {
    "Category": [
        "Electronics/Mobiles/Smartphones",
        "Home/Kitchen/Appliances",
        "Fashion/Men/Shirts",
        "Beauty/Makeup/Lipstick",
        "Books/Fiction/Novels"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["MainCategory"] = df["Category"].str.split("/").str[0]

print(df)
```

```python
# 5:
df = pd.DataFrame({
    "title": [
        "how to learn python",
        "best youtube tips",
        "data analytics tutorial",
        "python for beginners"
    ]
})

print(df)
```

```python
df["title"] = df["title"].str.upper().str.replace(" ", "_")
```
