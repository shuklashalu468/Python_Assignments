```python
# Task 2: Restaurant Menu Manager

# Menu stored in a dictionary
menu = {
    "Paneer Burger": {"price": 180, "category": "Snacks"},
    "Masala Dosa": {"price": 90, "category": "Breakfast"},
    "Veg Biryani": {"price": 220, "category": "Main Course"},
    "Margherita Pizza": {"price": 250, "category": "Main Course"},
    "Cold Coffee": {"price": 120, "category": "Beverages"},
    "Gulab Jamun": {"price": 80, "category": "Dessert"}
}


# Function 1: View all menu items
def view_all_items():
    print("\n----- Restaurant Menu -----")
    print("No.  Dish Name           Price    Category")
    print("---------------------------------------------")

    for number, (dish, details) in enumerate(menu.items(), 1):
        print(f"{number}.   {dish:<20} Rs {details['price']:<5} {details['category']}")


# Function 2: Filter items by category
def filter_by_category():
    category = input("Enter category: ").strip().lower()

    print(f"\n----- {category.title()} Items -----")

    found = False

    for dish, details in menu.items():
        if details["category"].lower() == category:
            print(f"{dish} - Rs {details['price']}")
            found = True

    if not found:
        print("No items found in this category.")


# Function 3: Search for a dish
def search_dish():
    dish_name = input("Enter dish name: ").strip().lower()

    for dish, details in menu.items():
        if dish.lower() == dish_name:
            print(f"{dish} - Rs {details['price']}")
            return

    print("Dish not found.")


# Main loop
while True:
    print("\n===== Restaurant Menu Manager =====")
    print("1. View all items")
    print("2. Filter items by category")
    print("3. Search for a dish")
    print("0. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        view_all_items()

    elif choice == "2":
        filter_by_category()

    elif choice == "3":
        search_dish()

    elif choice == "0":
        print("Thank you! Program ended.")
        break

    else:
        print("Invalid choice. Please try again.")
```

```python
# Task 3: Order History File Logger

import json

def load_orders():

    try:
        with open("orders.json", "r") as file:
            orders = json.load(file)

        if isinstance(orders, list):
            return orders

        elif isinstance(orders, dict):
            return [orders]

        else:
            return []

    except FileNotFoundError:
        return []

    except json.JSONDecodeError:
        return []

def save_orders(orders):

    with open("orders.json", "w") as file:
        json.dump(orders, file, indent=4)

def add_order():

    print("\n----- Add New Order -----")

    customer_name = input("Enter customer name: ")

    items_input = input("Enter items (comma-separated): ")

    items = [item.strip() for item in items_input.split(",")]

    try:
        total_amount = float(input("Enter total amount: "))

    except ValueError:
        print("Error: Total amount must be a number.")
        return

    status = input("Enter order status: ")

    orders = load_orders()

    new_order = {
        "customer_name": customer_name,
        "items": items,
        "total_amount": total_amount,
        "status": status
    }

    orders.append(new_order)

    save_orders(orders)

    print("\nOrder saved successfully!")


def view_orders():

    print("\n----- Order History -----")

    try:

        with open("orders.json", "r") as file:
            orders = json.load(file)

    except FileNotFoundError:

        print("No orders found.")
        return

    except json.JSONDecodeError:

        print("Error: orders.json contains invalid data.")
        return

    # If dictionary, convert it to list
    if isinstance(orders, dict):
        orders = [orders]

    if len(orders) == 0:

        print("No orders available.")
        return

    for i, order in enumerate(orders, start=1):

        print("\nOrder", i)
        print("-------------------------")
        print("Customer:", order["customer_name"])
        print("Items:", ", ".join(order["items"]))
        print("Total Amount: Rs", order["total_amount"])
        print("Status:", order["status"])

while True:

    print("\n==============================")
    print("   FOOD DELIVERY APP")
    print("==============================")
    print("1. Add Order")
    print("2. View Orders")
    print("0. Exit")
    print("==============================")

    choice = input("Enter your choice: ")


    if choice == "1":

        add_order()

    elif choice == "2":

        view_orders()

    elif choice == "0":

        print("\nThank you! Program ended.")
        break

    
    else:

        print("\nInvalid choice. Please enter 1, 2 or 0.")
```

```python
import csv
import os


# Rider class
class Rider:

    def __init__(self, rider_id, name, status="Available", total_deliveries=0):
        self.rider_id = rider_id
        self.name = name
        self.status = status
        self.total_deliveries = total_deliveries

    def assign_order(self, order_id):
        self.status = "On Delivery"
        print("Order", order_id, "assigned to", self.name)

    def complete_delivery(self):
        self.total_deliveries += 1
        self.status = "Available"
        print("Delivery completed by", self.name)

    def display_info(self):
        print("Rider ID:", self.rider_id)
        print("Name:", self.name)
        print("Status:", self.status)
        print("Total Deliveries:", self.total_deliveries)


# List to store rider objects
riders = []


# Load riders from CSV if file exists
if os.path.exists("riders.csv"):

    with open("riders.csv", "r") as file:

        reader = csv.DictReader(file)

        for row in reader:

            rider = Rider(
                int(row["rider_id"]),
                row["name"],
                row["status"],
                int(row["total_deliveries"])
            )

            riders.append(rider)

else:

    # Create at least three riders
    riders.append(Rider(1, "Rahul"))
    riders.append(Rider(2, "Amit"))
    riders.append(Rider(3, "Priya"))


# Main menu
while True:

    print("\n==============================")
    print("      DELIVERY RIDER SYSTEM")
    print("==============================")
    print("1. Display Riders")
    print("2. Assign Order")
    print("3. Complete Delivery")
    print("0. Exit")
    print("==============================")

    choice = input("Enter your choice: ")


    # Display riders
    if choice == "1":

        for rider in riders:

            print("\n----------------------")
            rider.display_info()


    # Assign order
    elif choice == "2":

        rider_id = int(input("Enter rider ID: "))
        order_id = input("Enter order ID: ")

        found = False

        for rider in riders:

            if rider.rider_id == rider_id:

                rider.assign_order(order_id)
                found = True
                break

        if not found:
            print("Rider not found.")


    # Complete delivery
    elif choice == "3":

        rider_id = int(input("Enter rider ID: "))

        found = False

        for rider in riders:

            if rider.rider_id == rider_id:

                rider.complete_delivery()
                found = True
                break

        if not found:
            print("Rider not found.")


    # Exit and save
    elif choice == "0":

        with open("riders.csv", "w", newline="") as file:

            writer = csv.writer(file)

            writer.writerow([
                "rider_id",
                "name",
                "status",
                "total_deliveries"
            ])

            for rider in riders:

                writer.writerow([
                    rider.rider_id,
                    rider.name,
                    rider.status,
                    rider.total_deliveries
                ])

        print("\nRider data saved successfully.")
        print("Program ended.")
        break


    else:

        print("Invalid choice. Please enter 1, 2, 3 or 0.")
```

```python
# SEC C
# Mini Project: Food Delivery Order Management Console

import json
import os

class Order:

    def __init__(
        self,
        order_id,
        customer_name,
        items,
        total_amount,
        status="Pending"
    ):
        self.order_id = order_id
        self.customer_name = customer_name
        self.items = items
        self.total_amount = total_amount
        self.status = status

    def to_dict(self):

        return {
            "order_id": self.order_id,
            "customer_name": self.customer_name,
            "items": self.items,
            "total_amount": self.total_amount,
            "status": self.status
        }

    @classmethod
    def from_dict(cls, data):

        return cls(
            data["order_id"],
            data["customer_name"],
            data["items"],
            data["total_amount"],
            data.get("status", "Pending")
        )

orders = []

def load_orders():

    global orders

    try:

        with open("orders.json", "r") as file:

            data = json.load(file)

        # Check JSON is a list
        if not isinstance(data, list):

            print("Invalid orders.json format.")
            orders = []
            return

        orders = []

        for order_data in data:

            required_keys = [
                "order_id",
                "customer_name",
                "items",
                "total_amount"
            ]

            if all(key in order_data for key in required_keys):

                order = Order.from_dict(order_data)

                orders.append(order)

            else:

                print("Skipped an invalid order in orders.json.")

        print("Orders loaded successfully.")


    except FileNotFoundError:

        print("orders.json not found.")
        print("Starting with an empty order list.")

        orders = []


    except json.JSONDecodeError:

        print("Error: orders.json contains invalid JSON.")

        print("Starting with an empty order list.")

        orders = []

def save_orders():

    data = []

    for order in orders:

        data.append(order.to_dict())

    with open("orders.json", "w") as file:

        json.dump(
            data,
            file,
            indent=4
        )

def generate_order_id():

    if len(orders) == 0:

        return 1

    max_id = 0

    for order in orders:

        if order.order_id > max_id:

            max_id = order.order_id

    return max_id + 1

def place_order():

    print("\n================================")
    print("        PLACE NEW ORDER")
    print("================================")

    while True:

        customer_name = input(
            "Enter customer name: "
        ).strip()

        if customer_name == "":

            print("Error: Customer name cannot be empty.")

        else:

            break

    while True:

        items_input = input(
            "Enter items (comma-separated): "
        ).strip()

        if items_input == "":

            print("Error: Please enter at least one item.")

        else:

            items = []

            for item in items_input.split(","):

                item = item.strip()

                if item != "":

                    items.append(item)

            if len(items) == 0:

                print("Error: Please enter valid items.")

            else:

                break

    while True:

        try:

            total_amount = float(
                input("Enter total amount: ")
            )

            if total_amount < 0:

                print(
                    "Error: Amount cannot be negative."
                )

            else:

                break

        except ValueError:

            print(
                "Error: Please enter a numeric amount."
            )

    order_id = generate_order_id()

    new_order = Order(
        order_id,
        customer_name,
        items,
        total_amount
    )

    orders.append(new_order)

    save_orders()


    print("\nOrder placed successfully!")
    print("Order ID:", order_id)
    print("Status:", new_order.status)

def view_all_orders():

    print("\n============================================================")
    print("                    ALL ORDERS")
    print("============================================================")


    if len(orders) == 0:

        print("No orders available.")

        return
        
    print(
        f"{'ID':<6}"
        f"{'Customer':<20}"
        f"{'Items':<8}"
        f"{'Amount':<15}"
        f"{'Status':<15}"
    )

    print("-" * 64)


    for order in orders:

        if order.status.lower() == "delivered":

            print(
                f"*** {order.order_id:<3}"
                f"{order.customer_name:<20}"
                f"{len(order.items):<8}"
                f"Rs {order.total_amount:<11.2f}"
                f"{order.status:<15} ***"
            )

        else:

            print(
                f"{order.order_id:<6}"
                f"{order.customer_name:<20}"
                f"{len(order.items):<8}"
                f"Rs {order.total_amount:<11.2f}"
                f"{order.status:<15}"
            )


    print("-" * 64)

    print("*** = Delivered Order ***")

def search_order():

    print("\n================================")
    print("        SEARCH ORDER")
    print("================================")


    try:

        order_id = int(
            input("Enter Order ID: ")
        )

    except ValueError:

        print("Error: Order ID must be a number.")

        return


    for order in orders:

        if order.order_id == order_id:

            print("\nOrder Found!")
            print("----------------------------")
            print("Order ID:", order.order_id)
            print("Customer:", order.customer_name)
            print("Items:", ", ".join(order.items))
            print("Item Count:", len(order.items))
            print(
                "Total Amount: Rs",
                f"{order.total_amount:.2f}"
            )
            print("Status:", order.status)
            print("----------------------------")

            return

    print("No order found with ID:", order_id)

load_orders()

while True:

    print("\n")
    print("========================================")
    print("   FOOD DELIVERY ORDER MANAGEMENT")
    print("========================================")
    print("1. Place New Order")
    print("2. View All Orders")
    print("3. Search Order by ID")
    print("4. Exit")
    print("========================================")


    choice = input(
        "Enter your choice: "
    ).strip()


    # Place Order

    if choice == "1":

        place_order()


    # View Orders

    elif choice == "2":

        view_all_orders()


    # Search Order

    elif choice == "3":

        search_order()


    # Exit

    elif choice == "4":

        print("\nThank you for using the Food Delivery System!")

        break


    # Invalid Choice

    else:

        print(
            "Invalid choice. Please enter 1, 2, 3 or 4."
        )
```
