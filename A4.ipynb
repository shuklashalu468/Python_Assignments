```python
product_name = input("Enter the product name: ") 

print("Upper case :", product_name.upper())
print("Lower case :", product_name.lower())
```

```python
def clean_brand_name(name):
    name = name.strip()
    name = name.replace("-", " ")
    return name

brand = " Nothing-3a "
print("Original: ", brand)
print("Cleaned: ", clean_brand_name(brand))
```

```python
phone = "Apple iphone 14 Pro Max"

split_index = phone.find(" ")

brand = phone[:split_index]
model = phone[split_index + 1:]

print("Brand:", brand)
print("Model:", model)
```

```python
def format_product_display(name, price):
    return f"{name} - Rs{price}"

result = format_product_display("Boat Earbuds", 1299)
print(result)
                            
```

```python
products = [' mi-Band 5 ', ' SAMSUNG-Galaxy ', ' realme-Book ']

cleaned_products = []

for product in products:
    product = product.strip()          # Remove leading/trailing spaces
    product = product.replace("-", " ")  # Replace hyphen with space
    product = product.title()          # Convert each word to Title Case
    cleaned_products.append(product)

print("Cleaned Product List:")
print(cleaned_products)
```
