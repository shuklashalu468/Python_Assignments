```python
import numpy as np
likes = np.array([120, 250, 180, 300, 450, 200, 350])
comments = np.array([10, 20, 15, 25, 40, 18, 30])

engagement = likes + comments

average_engagement = engagement.mean()

print("Average engagement per post:", average_engagement)
```

```python
prices = np.array([250, 300, 450, 200, 350])
discounts = np.array([50, 30, 100, 20, 70])

final_price = prices - discounts

print(final_price)
```

```python
scores = np.array([175, 195, 160, 210, 180])

result = scores > 180

print(result)
```

```python
paytm = np.array([1, 0, 1, 0, 1, 0])
phonepe = np.array([0, 1, 1, 0, 0, 1])

result = np.logical_or(paytm, phonepe)

print(result)
```

```python
steps = np.array([5000, 7000, 4500, 8000, 6000, 7500, 5500])

# Broadcasting: add 500 to every day's steps
bonus_steps = steps + 500

print("Steps after bonus:", bonus_steps)

# Aggregate operation: calculate total
total_steps = bonus_steps.sum()

print("Total steps:", total_steps)
```
