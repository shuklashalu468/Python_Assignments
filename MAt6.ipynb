```python
# 1) Use plt.scatter() to plot a graph of 'hours spent on Instagram per day' vs 'number of posts uploaded per month' for at least 
# 10 fictional users.

import matplotlib.pyplot as plt
import numpy as np

hours = [1, 2, 2.5, 3, 3.5, 4, 4.5, 5, 6, 7]
posts = [2, 4, 5, 7, 8, 10, 12, 14, 17, 20]

plt.scatter(hours, posts)

plt.xlabel("Hours Spent on Instagram per Day")
plt.ylabel("Number of Posts Uploaded per Month")
plt.title("Instagram Usage vs Monthly Posts")

plt.show()
```

```python
# 2) Create a scatter plot showing 'amount spent on Zomato orders' vs 'number of orders placed' for 12 friends, and label each point
# with the friend's name.<br><br><em><strong>Hint:</strong> Use the 'annotate' function in matplotlib to add labels.</em>

names = ["Aman", "Riya", "Rahul", "Priya", "Neha", "Karan",
         "Anjali", "Rohit", "Sneha", "Arjun", "Pooja", "Vikas"]

amount_spent = [500, 1200, 800, 1500, 700, 2000,
                1100, 900, 1800, 600, 1300, 1000]

orders = [3, 6, 4, 8, 3, 10,
          5, 4, 9, 2, 7, 5]

plt.scatter(amount_spent, orders)

for i in range(len(names)):
    plt.annotate(
        names[i],
        (amount_spent[i], orders[i]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("Amount Spent on Zomato Orders (₹)")
plt.ylabel("Number of Orders Placed")
plt.title("Zomato Spending vs Number of Orders")

plt.show()
```

```python
# 3) Add a trend line to your Zomato scatter plot by calculating the best-fit line using numpy's polyfit(), then plot it on top of the
# scatter plot.<br><br><em><strong>Hint:</strong> Use np.polyfit() to get the slope and intercept, then plot the line with plt.plot()

names = ["Aman", "Riya", "Rahul", "Priya", "Neha", "Karan",
         "Anjali", "Rohit", "Sneha", "Arjun", "Pooja", "Vikas"]

amount_spent = [500, 1200, 800, 1500, 700, 2000,
                1100, 900, 1800, 600, 1300, 1000]

orders = [3, 6, 4, 8, 3, 10,
          5, 4, 9, 2, 7, 5]

plt.scatter(amount_spent, orders)

for i in range(len(names)):
    plt.annotate(names[i], (amount_spent[i], orders[i]))

slope, intercept = np.polyfit(amount_spent, orders, 1)

line = slope * np.array(amount_spent) + intercept

plt.plot(amount_spent, line)

plt.xlabel("Amount Spent on Zomato Orders (₹)")
plt.ylabel("Number of Orders Placed")
plt.title("Zomato Spending vs Number of Orders")

plt.show()
```

```python
# 4) Given a dataset of 'YouTube watch hours' and 'Spotify listening hours' for 15 users, use numpy polyfit to fit a regression line,
# plot both the scatter and the trend line, and print the equation of the line on the plot.

youtube_hours = [2, 3, 4, 5, 6, 2.5, 3.5, 4.5, 5.5, 6.5,
                 7, 3, 4, 5, 7.5]

spotify_hours = [1, 2, 2.5, 3.5, 4, 2, 3, 3.5, 4.5, 5,
                 5.5, 2.5, 3, 4, 6]

plt.scatter(youtube_hours, spotify_hours)

slope, intercept = np.polyfit(youtube_hours, spotify_hours, 1)

line = slope * np.array(youtube_hours) + intercept

plt.plot(youtube_hours, line)

equation = f"y = {slope:.2f}x + {intercept:.2f}"

plt.text(2, 5.5, equation)

plt.xlabel("YouTube Watch Hours")
plt.ylabel("Spotify Listening Hours")
plt.title("YouTube Watch Hours vs Spotify Listening Hours")

plt.show()
```
