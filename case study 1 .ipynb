```python
# 1) Download the Kaggle Bank Marketing Dataset and use Pandas to load it into a DataFrame named bank_df, then display the first 10 rows
# and the DataFrame's info().

import pandas as pd

bank_df = pd.read_csv("bank-full.csv", sep=";")

print(bank_df.head(10))

bank_df.info()
```

```python
# Use Seaborn's countplot to visualize how many customers in the dataset subscribed to a term deposit (column 'y'), and display the plot
# with appropriate axis labels.

import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(
    data=bank_df,
    x="y"
)

plt.xlabel("Term Deposit Subscription")
plt.ylabel("Number of Customers")
plt.title("Customers Who Subscribed to a Term Deposit")

plt.show()
```

```python
# Analyze the relationship between customer age and term deposit subscription by creating a Seaborn boxplot with 'y' on the x-axis and
# 'age' on the y-axis.<br><br><em><strong>Hint:</strong> This helps you see if certain age groups are more likely to subscribe.</em>

sns.boxplot(
    data=bank_df,
    x="y",
    y="age"
)

plt.xlabel("Term Deposit Subscription")
plt.ylabel("Customer Age")
plt.title("Customer Age vs Term Deposit Subscription")

plt.show()
```

```python
# Create a bar plot using Seaborn to show the percentage of customers who subscribed to a term deposit for each education level
#     (column 'education').<br><br><em><strong>Hint:</strong> Calculate the percentage for each group before plotting.</em>

education_percentage = (
    bank_df.groupby("education")["y"]
    .apply(lambda x: (x == "yes").mean() * 100)
    .reset_index(name="subscription_percentage")
)

print(education_percentage)

sns.barplot(
    data=education_percentage,
    x="education",
    y="subscription_percentage"
)

plt.xlabel("Education Level")
plt.ylabel("Subscribed Customers (%)")
plt.title("Term Deposit Subscription Percentage by Education")

plt.xticks(rotation=45)

plt.show()
```

```python
# Use ChatGPT or Copilot to suggest two additional visualizations that could help a fintech startup like Cred understand which users
# are most likely to buy a new savings product, then implement one of those visualizations using Seaborn.

job_subscription = (
    bank_df.groupby("job")["y"]
    .apply(lambda x: (x == "yes").mean() * 100)
    .reset_index(name="subscription_percentage")
)

print(job_subscription)

sns.barplot(
    data=job_subscription,
    x="job",
    y="subscription_percentage"
)

plt.xlabel("Job")
plt.ylabel("Subscription Rate (%)")
plt.title("Term Deposit Subscription Rate by Job")

plt.xticks(rotation=45)

plt.show()
```
