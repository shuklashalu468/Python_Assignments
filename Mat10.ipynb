```python
# 1) Create a line chart using Matplotlib to show the number of daily active users for a music streaming app (like Spotify) over 12 
# months, and use plt.annotate() to label the month with the highest user count.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

active_users = [1200, 1350, 1500, 1450, 1600, 1750,
                1900, 1850, 2100, 2300, 2200, 2500]

plt.figure(figsize=(10, 5))

plt.plot(months, active_users, marker="o")

max_users = max(active_users)
max_month = months[active_users.index(max_users)]

plt.annotate(
    f"Highest: {max_users}",
    xy=(max_month, max_users),
    xytext=(0, 20),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)

plt.xlabel("Month")
plt.ylabel("Daily Active Users")
plt.title("Monthly Daily Active Users - Music Streaming App")

plt.grid(True)
plt.show()
```

```python
# 2) Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using
# plt.annotate() to highlight the month with the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw
# the arrow pointing to the bar.</em>

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

orders = [120, 150, 110, 180, 160, 200,
          140, 130, 170, 190, 125, 210]

plt.figure(figsize=(10, 5))

plt.bar(months, orders)

min_orders = min(orders)
min_month = months[orders.index(min_orders)]

plt.annotate(
    f"Lowest: {min_orders}",
    xy=(min_month, min_orders),
    xytext=(0, 40),
    textcoords="offset points",
    ha="center",
    arrowprops=dict(arrowstyle="->")
)

plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.title("Monthly Food Delivery Orders")

plt.show()
```

```python
# 3) Create a pie chart representing the percentage of different payment methods (UPI, Card, Wallet, Cash) used on an e-commerce app 
# (like Flipkart), and add a legend positioned at the lower right corner of the plot.

payment_methods = ["UPI", "Card", "Wallet", "Cash"]
usage = [45, 30, 15, 10]

plt.figure(figsize=(7, 5))

plt.pie(usage, labels=payment_methods, autopct="%1.1f%%")

plt.title("Payment Methods Used on E-commerce App")

plt.legend(loc="lower right")

plt.show()
```

```python
# 4) Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use 
# plt.annotate() to add a text box highlighting the movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox'
# parameter in plt.annotate() to create a visible text box.</em>

movies = ["Movie A", "Movie B", "Movie C", "Movie D", "Movie E",
          "Movie F", "Movie G", "Movie H"]

ratings = [7.5, 8.2, 9.1, 7.8, 8.7, 7.2, 8.5, 7.9]
reviews = [1200, 2500, 1800, 3000, 2200, 1500, 2700, 1900]

plt.figure(figsize=(9, 5))

plt.scatter(reviews, ratings)

max_rating = max(ratings)
max_movie = movies[ratings.index(max_rating)]
max_reviews = reviews[ratings.index(max_rating)]

plt.annotate(
    f"{max_movie}\nRating: {max_rating}",
    xy=(max_reviews, max_rating),
    xytext=(20, 20),
    textcoords="offset points",
    bbox=dict(boxstyle="round,pad=0.5", edgecolor="black"),
    arrowprops=dict(arrowstyle="->")
)

plt.xlabel("Number of Reviews")
plt.ylabel("Movie Rating")
plt.title("Movie Ratings vs Number of Reviews")

plt.show()
```
