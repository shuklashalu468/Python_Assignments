```python
# Recursive function to reverse a string
def reverse_string(s):
    if len(s) == 0:
        return s
    return reverse_string(s[1:]) + s[0]

# Test the function
result = reverse_string("hello")
print("Reversed String:", result)
```

```python
# Recursive function to calculate the total playlist duration
def sum_playlist_durations(durations):
    # Base case
    if len(durations) == 0:
        return 0

    # Recursive case
    return durations[0] + sum_playlist_durations(durations[1:])

# List of song durations (in seconds)
playlist = [180, 210, 240, 150]

# Call the function
total_duration = sum_playlist_durations(playlist)

print("Total Playlist Duration:", total_duration, "seconds")
```

```python
def count_likes(posts):
    total = posts["likes"]

    for reply in posts.get("replies", []):
        total += count_likes(reply)

    return total


posts = {
    "likes": 100,
    "replies": [
        {
            "likes": 50,
            "replies": [
                {
                    "likes": 20,
                    "replies": []
                }
            ]
        },
        {
            "likes": 30,
            "replies": []
        }
    ]
}

print("Total likes:", count_likes(posts))
```

```python
# Global variable
app_status = "Offline"

print("Before function call:")
print("App Status:", app_status)


def change_status():
    # Local variable
    user_status = "Online"

    print("\nInside function:")
    print("User Status:", user_status)
    print("App Status:", app_status)


change_status()

print("\nAfter function call:")
print("App Status:", app_status)

# user_status cannot be accessed here because it was local to the function
```
