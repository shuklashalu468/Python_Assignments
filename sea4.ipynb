```python
#1) Load a CSV file containing Flipkart product prices and use sns.boxplot() to visualize the price distribution.

import pandas as pd

data = {
    "product": [
        "Mobile", "Laptop", "Headphones", "Smartwatch", "Tablet",
        "Keyboard", "Mouse", "Monitor", "Speaker", "Camera"
    ],
    "price": [
        15000, 45000, 2500, 5000, 22000,
        1500, 800, 12000, 3500, 28000
    ]
}

df = pd.DataFrame(data)

df.to_csv("flipkart_products.csv", index=False)

print("CSV file created successfully!")

df = pd.read_csv("flipkart_products.csv")

print(df)

import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(data=df, y="price")

plt.ylabel("Product Price (₹)")
plt.title("Distribution of Flipkart Product Prices")

plt.show()
```

```python
# 2) Use sns.violinplot() to compare the order amounts for two different Zomato restaurant categories (e.g., 'Fast Food' vs 'Desserts')
# from a sample dataset.

import pandas as pd

data = {
    "category": [
        "Fast Food", "Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Fast Food", "Fast Food", "Fast Food", "Fast Food", "Fast Food",
        "Desserts", "Desserts", "Desserts", "Desserts", "Desserts",
        "Desserts", "Desserts", "Desserts", "Desserts", "Desserts"
    ],
    
    "order_amount": [
        250, 300, 450, 350, 500,
        280, 400, 550, 320, 600,
        150, 200, 250, 300, 180,
        220, 350, 280, 400, 200
    ]
}

df = pd.DataFrame(data)

print(df)

import seaborn as sns
import matplotlib.pyplot as plt

sns.violinplot(
    data=df,
    x="category",
    y="order_amount"
)

plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (₹)")
plt.title("Order Amount Distribution by Zomato Category")

plt.show()
```

```python
# 3) Given an array of IPL match ticket prices, plot a boxplot using sns.boxplot() and identify any outliers in the data.
    # <br><br><em><strong>Hint:</strong> Outliers are points that appear outside the whiskers of the boxplot.</em>

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

ticket_prices = [
    500, 700, 800, 1000, 1200,
    900, 1100, 750, 950, 1300,
    1000, 850, 1250, 700, 1500,
    1800, 2000, 5000
]

df = pd.DataFrame({
    "ticket_price": ticket_prices
})

print(df)

sns.boxplot(data=df, y="ticket_price")

plt.ylabel("Ticket Price (₹)")
plt.title("IPL Match Ticket Price Distribution")

plt.show()

Q1 = df["ticket_price"].quantile(0.25)
Q3 = df["ticket_price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["ticket_price"] < lower_limit) |
    (df["ticket_price"] > upper_limit)
]

print("Outliers:")
print(outliers)
```
