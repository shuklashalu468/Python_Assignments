```python
import numpy as np

zomato = np.array([4.5, 4.2, 3.8, 4.7, 4.0])
swiggy = np.array([4.1, 3.9, 4.6, 4.3, 4.8])

ratings = np.concatenate((zomato, swiggy))

print(ratings)
```

```python
# Messages sent by 3 users for 7 days
chat = np.array([
    [20, 15, 25, 10, 18, 22, 30],  # User 1
    [5, 8, 6, 4, 7, 5, 9],         # User 2
    [15, 20, 18, 22, 16, 19, 25]   # User 3
])

# Add a new user
new_user = np.array([10, 12, 15, 8, 11, 14, 13])

chat = np.insert(chat, 3, new_user, axis=0)

print("After adding new user:")
print(chat)

# Find the user with the least total messages
totals = np.sum(chat, axis=1)
least_user = np.argmin(totals)

# Remove that user
chat = np.delete(chat, least_user, axis=0)

print("\nAfter removing least active user:")
print(chat)
```

```python
views = np.array([1000, 2000, 3000, 4000])

# Create a view and a copy
view = views.view()
copy = views.copy()

# Modify the first element
view[0] = 9999
copy[0] = 8888

print("Original array:", views)
print("View:", view)
print("Copy:", copy)
```
