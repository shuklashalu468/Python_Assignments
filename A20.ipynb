```python
class Product:
    def __init__(self, price):
        self._price = price

    # Getter
    def get_price(self):
        return self._price

    # Setter
    def set_price(self, price):
        if price < 0:
            raise ValueError("Price cannot be negative")
        self._price = price


# Create object
product1 = Product(500)

# Get price
print("Current Price: ₹", product1.get_price())

# Update price
product1.set_price(700)
print("Updated Price: ₹", product1.get_price())

# Try to set negative price
product1.set_price(-100)
```

```python
class Playlist:
    def __init__(self):
        self._songs = []

    # Add a song
    def add_song(self, song):
        self._songs.append(song)

    # Remove a song
    def remove_song(self, song):
        if song in self._songs:
            self._songs.remove(song)
        else:
            print("Song not found")

    # Get the current list of songs
    def get_songs(self):
        return self._songs.copy()


# Create a Playlist object
playlist = Playlist()

# Add songs
playlist.add_song("Kesariya")
playlist.add_song("Apna Bana Le")
playlist.add_song("Chaleya")

print("Playlist:", playlist.get_songs())

# Remove a song
playlist.remove_song("Apna Bana Le")

print("Updated Playlist:", playlist.get_songs())
```

```python
from abc import ABC, abstractmethod


# Abstract class
class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


# UPI subclass
class UPI(PaymentMethod):

    def pay(self, amount):
        print(f"Processing ₹{amount} payment through UPI")


# Credit Card subclass
class CreditCard(PaymentMethod):

    def pay(self, amount):
        print(f"Processing ₹{amount} payment through Credit Card")


# Create objects
upi_payment = UPI()
card_payment = CreditCard()

# Make payments
upi_payment.pay(500)
card_payment.pay(1000)
```
