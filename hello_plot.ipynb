```python
# 1) Install Matplotlib in your Python environment and write a script named hello_plot.py that creates a simple line plot showing
# the number of daily steps you walked over the past 7 days.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [5000, 6500, 7200, 6000, 8000, 9500, 7500]

plt.plot(days, steps, marker="o")

plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.title("Daily Steps Over 7 Days")

plt.show()
```

```python
# 3) Write a Python script that creates a Matplotlib figure with two Axes: one showing your daily Spotify listening hours for a week
# and the other showing your daily YouTube viewing hours for the same week. Display both plots in a single figure.

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify_hours = [2, 3, 2.5, 4, 3, 5, 4]
youtube_hours = [1, 2, 1.5, 2.5, 3, 4, 3.5]

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].plot(days, spotify_hours, marker="o")
ax[0].set_title("Daily Spotify Listening")
ax[0].set_xlabel("Day")
ax[0].set_ylabel("Hours")

ax[1].plot(days, youtube_hours, marker="o")
ax[1].set_title("Daily YouTube Viewing")
ax[1].set_xlabel("Day")
ax[1].set_ylabel("Hours")

plt.tight_layout()
plt.show()
```

```python
# 4) Add axis labels, a title, and a grid to your plot from the previous task to clearly communicate the information, mimicking the style
# of analytics dashboards in apps like Instagram Insights.<br><br><em><strong>Hint:</strong> Use set_xlabel(), set_ylabel(), set_title(),
# and grid() methods on the Axes object.</em>

fig, ax = plt.subplots(2, 2, figsize=(8,6))

ax[0,0].plot([1,2,3,4,5])
ax[0,0].set_xlabel("Days")
ax[0,0].set_ylabel("Values")
ax[0,0].set_title("Line Plot")
ax[0,0].grid()

ax[0,1].bar([1,2,3,4,5], [2,3,5,7,8])
ax[0,1].set_xlabel("Categories")
ax[0,1].set_ylabel("Values")
ax[0,1].set_title("Bar Plot")
ax[0,1].grid()

ax[1,0].scatter([1,2,3,4,5], [3,5,8,11,15])
ax[1,0].set_xlabel("X Values")
ax[1,0].set_ylabel("Y Values")
ax[1,0].set_title("Scatter Plot")
ax[1,0].grid()

ax[1,1].hist([1,2,3,4,5,8,11,15])
ax[1,1].set_xlabel("Values")
ax[1,1].set_ylabel("Frequency")
ax[1,1].set_title("Histogram")
ax[1,1].grid()

plt.tight_layout()
plt.show()
```
