```python
songs = [
    "Perfect",
    "Shape of You",
    "Believer",
    "Tum Hi Ho",
    "Kesariya"
]

with open("my_playlist.txt", "w") as file:
    for song in songs:
        file.write(song + "\n")

print("Playlist created successfully!")
```

```python
with open("my_playlist.txt", "r") as file:
    for song in file:
        print(song.strip().upper())
```
