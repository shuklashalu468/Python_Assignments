```python
# Function to calculate the final price
def calculate_final_price(price, discount_rate):
    final_price = price - (price * discount_rate)
    return final_price

# Test the function
result = calculate_final_price(1200, 0.15)

print("Final Price:", result)
```

```python
# Function with a default argument
def get_delivery_charge(amount, city="Ahmedabad"):
    if city == "Ahmedabad":
        return 30
    else:
        return 50

# Call without city argument (uses default city)
charge1 = get_delivery_charge(500)
print("Delivery Charge:", charge1)

# Call with city argument
charge2 = get_delivery_charge(500, "Mumbai")
print("Delivery Charge:", charge2)
```

```python
# Function with a default discount
def format_coupon_message(username, discount=10):
    return f"Hi {username}, you get {discount}% off!"

# Test with the default discount
message1 = format_coupon_message("Rahul")

# Test with a custom discount
message2 = format_coupon_message("Priya", 20)

print(message1)
print(message2)
```

```python
# Function with a default discount rate
def apply_discount(price, rate=0.10):
    final_price = price - (price * rate)
    return final_price

# Call the function with only the price argument
result = apply_discount(1000)

print("Final Price:", result)
```

```python
# Function to calculate cashback
def calculate_cashback(amount, cashback_rate=0.05):
    cashback = amount * cashback_rate
    return cashback

# Zomato order with default cashback rate (5%)
zomato_cashback = calculate_cashback(500)

# Flipkart order with custom cashback rate (7%)
flipkart_cashback = calculate_cashback(2000, 0.07)

print("Zomato Cashback: Rs.", zomato_cashback)
print("Flipkart Cashback: Rs.", flipkart_cashback)
```
