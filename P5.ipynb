```python
import pandas as pd
```

```python
# 1:
data = {
    "player_name": [
        "Virat Kohli",
        "Rohit Sharma",
        "MS Dhoni",
        "Jasprit Bumrah",
        "Shubman Gill",
        "Ravindra Jadeja",
        "Hardik Pandya",
        "KL Rahul"
    ],
    "team": [
        "RCB",
        "MI",
        "CSK",
        "MI",
        "GT",
        "CSK",
        "MI",
        "LSG"
    ],
    "age": [
        37,
        39,
        None,
        32,
        27,
        None,
        32,
        34
    ]
}
df = pd.DataFrame(data)

df.to_csv("ipl_players.csv", index=False)

df = pd.read_csv("ipl_players.csv")

print("IPL Players Data:")
print(df)

missing_age = df[df["age"].isna()]

print("Players with missing age:")
print(missing_age)
```

```python
# 2:
data = {
    "order_id": [101, 102, 103, 104, 105, 106],
    "customer_name": [
        "Rahul",
        "Priya",
        "Aman",
        "Neha",
        "Riya",
        "Karan"
    ],
       "delivery_rating": [5, None, 4, None, 3, 5]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
df = df.dropna(subset=["delivery_rating"])
print("Cleaned DataFrame:")
print(df)
```

```python
# 3:
data = {
    "product_name": [
        "Mobile",
        "Laptop",
        "Headphones",
        "Smart Watch",
        "Tablet",
        "Keyboard"
    ],
    "rating": [4.5, 4.0, None, 3.5, 5.0, None]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

mean_rating = df["rating"].mean()

df["rating"] = df["rating"].fillna(mean_rating)

print("Updated DataFrame:")
print(df)
```

```python
# 4:
df = pd.read_csv("Spotify_playlist.csv")

print("Original DataFrame:")
print(df)

median_duration = df["duration"].median()

df["duration"] = df["duration"].fillna(median_duration)

print("Cleaned DataFrame:")
print(df)

df.to_csv("Spotify_playlist_cleaned.csv", index=False)

print("Cleaned CSV file saved successfully!")
```

```python
# 5:
data = {
    "username": ["user1", "user2", "user3", "user4", "user5"],
    "followers": [1200, None, 5400, None, 8500]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

df["followers"] = df["followers"].fillna(0)

print("Updated Data:")
print(df)
```
