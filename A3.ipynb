```python
# 1. Declare four variables in Python: one integer (number of followers), one float (average rating), one string (your favorite app's name), and one boolean (is_premium_user). Print each variable and its type using the type() function.

followers = 4000
average_rating = 5.5
favorite_app = "Instagram"
is_premium_user = True

print("Followers :", followers)
print("Data Type:", type(followers))

print("Average Rating :", average_rating)
print("Data Type:", type(average_rating))

print("Favorite App :", favorite_app)
print("Data Type:", type(favorite_app))

print("Is Premium User :", is_premium_user)
print("Data Type:", type(is_premium_user))
```

```python
#2. Write a Python program that takes a user's input for the price of a Zomato order as a string, converts it to a float using type casting, adds 18% GST, and prints the final bill amount.

price = input("Enter the zomato order price: ")
price = float(price)
final_bill = price + (price * 18/100)
print(f"final bill amount(including 18% gst):", final_bill)
```

```python
# 3. Given a list of strings representing product prices from Flipkart, like ['199.99', '299.50', '150'], convert all to floats and calculate the total cart value.

prices = ['199.99', '299.5', '150']
float_prices = [float(price) for price in prices]
total = sum(float_prices)
print("Prices as floats:" ,float_prices)
print("Total cart value:" ,total )
```

```python
# 4 . function to check discount eligibility
def is_discount_applicable(order_amount):
    return order_amount > 500

print(is_discount_applicable(450))
print(is_discount_applicable(750))
```

```python
#5 in list comprehension method

ratings = ['4.5', '3.0', '5', '4.2']
float_ratings = [float(rating) for rating in ratings]
highest_rating = max(float_ratings)
print("Ratings:", float_ratings)
print("Highest Ratings:", highest_rating)
```

```python
# using a for loop

ratings = ['4.5', '3.0', '5', '4.2']
float_ratings = []
for rating in ratings :
    float_ratings.append(float(rating))
highest_rating = max(float_ratings)
print("Ratings:", float_ratings)
print("Highest Ratings:", highest_rating)
```
