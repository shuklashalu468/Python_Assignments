```python
# 1:
import pandas as pd

january_orders = pd.DataFrame({
    "order_id": [101, 102, 103],
    "customer": ["Rahul", "Priya", "Aman"],
    "food": ["Pizza", "Biryani", "Burger"],
    "amount": [300, 250, 200]
})

print(january_orders)
```

```python
february_orders = pd.DataFrame({
    "order_id": [104, 105, 106],
    "customer": ["Neha", "Riya", "Karan"],
    "food": ["Momos", "Pasta", "Dosa"],
    "amount": [150, 280, 180]
})

print(february_orders)
```

```python
all_orders = pd.concat(
    [january_orders, february_orders],
    ignore_index=True
)

print(all_orders)
```

```python
# 2:

products = pd.DataFrame({
    "product_id": [101, 102, 103, 104],
    "name": ["Mobile", "Laptop", "Headphones", "Watch"],
    "category": ["Electronics", "Electronics", "Electronics", "Accessories"]
})

ratings = pd.DataFrame({
    "rating": [5, 4, 4, 3]
})

print("Products:")
print(products)

print("Ratings:")
print(ratings)
```

```python
result = pd.concat(
    [products, ratings],
    axis=1
)

print(result)
```

```python
# 3:

data = {
    "user": [
        "Rahul", "Rahul", "Rahul", "Rahul", "Rahul", "Rahul", "Rahul",
        "Priya", "Priya", "Priya", "Priya", "Priya", "Priya", "Priya",
        "Aman", "Aman", "Aman", "Aman", "Aman", "Aman", "Aman"
    ],
    "day": [
        "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday",
        "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday",
        "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"
    ],
    "streams": [
        120, 150, 130, 180, 200, 250, 220,
        200, 180, 220, 240, 260, 300, 280,
        100, 120, 110, 140, 160, 180, 150
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = df.pivot(
    index="user",
    columns="day",
    values="streams"
)

print(result)
```

```python
# 4:

data = {
    "restaurant": [
        "Dominos", "Dominos", "Dominos",
        "Burger King", "Burger King", "Burger King",
        "Biryani House", "Biryani House", "Biryani House"
    ],
    "month": [
        "January", "February", "March",
        "January", "February", "March",
        "January", "February", "March"
    ],
    "total_amount": [
        300, 400, 350,
        250, 300, 280,
        400, 450, 500
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
result = pd.pivot_table(
    df,
    index="restaurant",
    columns="month",
    values="total_amount",
    aggfunc="mean"
)

print(result)
```

```python
# 5:

jan = pd.DataFrame({
    "product": ["T-Shirt", "Jeans", "Shoes"],
    "sales": [10000, 15000, 20000]
})

feb = pd.DataFrame({
    "product": ["T-Shirt", "Jeans", "Shoes"],
    "sales": [12000, 18000, 22000]
})

mar = pd.DataFrame({
    "product": ["T-Shirt", "Jeans", "Shoes"],
    "sales": [15000, 20000, 25000]
})

jan.to_csv("sales_jan.csv", index=False)
feb.to_csv("sales_feb.csv", index=False)
mar.to_csv("sales_mar.csv", index=False)

print("CSV files created successfully!")
```

```python
jan = pd.read_csv("sales_jan.csv")
feb = pd.read_csv("sales_feb.csv")
mar = pd.read_csv("sales_mar.csv")

all_sales = pd.concat([jan, feb, mar], ignore_index=True)

print(all_sales)
```

```python
total_sales = all_sales.groupby("product")["sales"].sum()

print(total_sales)
```
