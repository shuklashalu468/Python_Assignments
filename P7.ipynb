```python
# 3:
import pandas as pd

data = {
    "PostID": [101, 102, 103, 104],
    "Likes": [1500, 25000, 2000000, 850],
    "Comments": [120, 3500, 1500000, 90]
}

df = pd.DataFrame(data)

print(df)
```

```python
def format_number(x):
    if x >= 1000000:
        return f"{x/1000000:.1f}M"
    elif x >= 1000:
        return f"{x/1000:.1f}K"
    else:
        return str(x)
```

```python
df[["Likes", "Comments"]] = df[["Likes", "Comments"]].map(format_number)
```

```python
# 4:
data = {
    "Restaurant": ["Dominos", "Burger King", "Biryani House", "Momos Point"],
    "Price": [300, 200, 250, 150],
    "Qty": [2, 3, 1, 4],
    "DeliveryCharge": [40, 30, 50, 20]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["FinalAmount"] = df.apply(
    lambda x: (x["Price"] * x["Qty"]) + x["DeliveryCharge"],
    axis=1
)

print(df)
```

```python
# 5:
data = {
    "Product": ["T-Shirt", "Jeans", "Shoes", "Kurti"],
    "Price": [800, 1500, 2000, 700],
    "Qty": [2, 1, 1, 3]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["GST"] = df.apply(
    lambda x: (x["Price"] * x["Qty"]) * 0.05,
    axis=1
)

print(df)
```
