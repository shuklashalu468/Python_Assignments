```python
import numpy as np

prices = np.array([299, 499, 799, 0, 1599, -1, 899])

print("Prices:", prices)
print("Data type:", prices.dtype)
```

```python
prices = np.array([299, 499, 799, 0, 1599, -1, 899])

# Find average of positive prices
average = prices[prices > 0].mean()

# Replace values <= 0 with the average
prices[prices <= 0] = average

print("Average:", average)
print("Updated prices:", prices)
```

```python
prices = np.array([299, 499, 799, 819, 1599, 819, 899])
quantities = np.array([2, 1, 3, 4, 2, 1, 5])

total = prices * quantities

print("Total bill for each item:", total)
```

```python
prices = np.array([299, 499, 799, 819, 1599, 819, 899])
quantities = np.array([2, 1, 3, 4, 2, 1, 5])

# Total bill for each item
total = prices * quantities

print("Minimum price:", prices.min())
print("Maximum price:", prices.max())
print("Average price:", prices.mean())
print("Total of prices:", prices.sum())
print("Total bill:", total.sum())
```

```python
prices = np.array([299, 499, 799, 819, 1599, 819, 899])

unique_prices = np.unique(prices)

print("Unique prices:", unique_prices)
print("Number of unique prices:", len(unique_prices))
```
