```python
import numpy as np

teams = np.array([
    "CSK", "MI", "RCB", "CSK", "KKR",
    "MI", "SRH", "RCB", "DC"
])

unique_teams = np.unique(teams)

print(unique_teams)
```

```python
ratings = np.array([4.5, 3.8, np.nan, 4.2, np.nan, 5.0, 3.5])

missing = np.isnan(ratings)

print("Missing values:", missing)
print("Number of missing ratings:", np.sum(missing))
```

```python
prices = np.array([50, 250, 750, 1200, 900, 1500, 80])

result = np.clip(prices, 100, 1000)

print(result)
```

```python
views = np.array([1000, 2500, np.nan, 5000, np.inf, 3500, 2000])

# Find the maximum finite value
max_value = np.max(views[np.isfinite(views)])

# Replace NaN with 0
views[np.isnan(views)] = 0

# Replace infinity with maximum finite value
views[np.isinf(views)] = max_value

print(views)
```

```python
numbers = np.array([11, 24, 37, 42, 55, 68, 71, 80])

even_indices = np.where(numbers % 2 == 0)

print("Even number indices:", even_indices)
```
