```python
# 1:
import pandas as pd

movies = pd.DataFrame({
    "title": ["3 Idiots", "Dangal", "Zindagi Na Milegi Dobara", "Taare Zameen Par", "PK", "Sholay"],
    "rating": [8.4, 8.3, 8.2, 8.3, 8.1, 8.1],
    "release_year": [2009, 2016, 2011, 2007, 2014, 1975]
})

movies.sort_values(by="rating", ascending=False)
```

```python
# 2:
ipl = pd.DataFrame({
    "team": ["CSK", "MI", "RCB", "KKR", "RR"],
    "points": [16, 18, 14, 12, 10],
    "net_run_rate": [0.45, 0.75, 0.25, 0.10, -0.20]
})

ipl.sort_index(ascending=False)
```

```python
# 3:
videos = pd.DataFrame({
    "video_title": ["Funny Cat Video", "New Song", "Cricket Highlights", "Travel Vlog", "Comedy Video"],
    "views": [5000000, 8000000, 3000000, 2000000, 6000000],
    "likes": [400000, 700000, 250000, 150000, 500000]
})

videos.index = [5, 8, 12, 15, 20]
```

```python
print("Before resetting index:")
print(videos)
```

```python
videos = videos.reset_index(drop=True)
```

```python
videos
```

```python
# 4:
restaurants = pd.DataFrame({
    "restaurant": ["Pizza Hut", "Dominos", "Biryani Blues", "Burger King", "Haldiram"],
    "rating": [4.5, 4.2, 4.3, 4.1, 4.4]
})

print("Original DataFrame:")
print(restaurants)
```

```python
restaurants = restaurants.reindex([2, 0, 1, 3, 4])
```

```python
print("After reindexing:")
print(restaurants)
```

```python
# 5:
products = pd.DataFrame({
    "product_name": ["Mobile", "Laptop", "Headphones", "Watch", "Tablet"],
    "price": [15000, 50000, 2000, 3000, 20000],
    "discount": [10, 30, 50, 20, 40]
})
```

```python
print("Original DataFrame:")
print(products)
```

```python
products = products.sort_values(by="discount", ascending=False)
```

```python
products = products.reset_index(drop=True)
```

```python
print("After sorting and resetting index:")
print(products)
```
