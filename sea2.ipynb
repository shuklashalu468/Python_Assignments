```python
# 1) Use sns.countplot() to visualize the number of orders placed by each payment method (e.g., UPI, Card, Cash)
# in a sample Swiggy orders DataFrame with at least 3 payment types.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "payment_method": [
        "UPI", "Card", "Cash", "UPI", "UPI",
        "Card", "Cash", "UPI", "Card", "UPI"
    ]
}

df = pd.DataFrame(data)

print(df)

sns.countplot(data=df, x="payment_method")

plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")
plt.title("Orders by Payment Method")

plt.show()
```

```python
# 2) Create a barplot using sns.barplot() to show the average delivery time (in minutes) for each food category
# (e.g., Pizza, Burger, Biryani, South Indian) from a Zomato-style dataset.<br><br><em><strong>Hint:</strong>
# Use the estimator parameter to show the mean delivery time for each category.</em>

data = {
    "food_category": [
        "Pizza", "Pizza", "Pizza",
        "Burger", "Burger", "Burger",
        "Biryani", "Biryani", "Biryani",
        "South Indian", "South Indian", "South Indian"
    ],
    "delivery_time": [
        30, 35, 25,
        20, 25, 30,
        40, 45, 35,
        25, 30, 20
    ]
}

df = pd.DataFrame(data)

print(df)

sns.barplot(
    data=df,
    x="food_category",
    y="delivery_time",
    estimator="mean"
)

plt.xlabel("Food Category")
plt.ylabel("Average Delivery Time (minutes)")
plt.title("Average Delivery Time by Food Category")

plt.show()
```

```python
# 3) Modify your barplot to display the total number of orders instead of the average by setting estimator to sum in sns.barplot().

sns.barplot(
    data=df,
    x="food_category",
    y="delivery_time",
    estimator="sum"
)

plt.xlabel("Food Category")
plt.ylabel("Total Delivery Time (minutes)")
plt.title("Total Delivery Time by Food Category")

plt.show()
```
