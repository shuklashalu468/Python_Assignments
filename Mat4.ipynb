```python
# 1) Use plt.pie() to visualize the percentage of time you spend on 5 different apps in a day (e.g., Instagram, YouTube, WhatsApp, Zomato,
# Spotify) and display the percentage values on each segment using the autopct parameter.

import matplotlib.pyplot as plt

apps = ["Instagram", "YouTube", "WhatsApp", "Zomato", "Spotify"]
time_spent = [3, 2.5, 2, 1, 1.5]

plt.pie(
    time_spent,
    labels=apps,
    autopct="%1.1f%%"
)

plt.title("Daily Time Spent on Different Apps")

plt.show()
```

```python
# 2) Create a pie chart showing the distribution of your monthly online spending across 4 categories: Food Delivery, Shopping, 
# Entertainment, and UPI Payments, and use the explode parameter to highlight the category with the highest spending.
# <br><br><em><strong>Hint:</strong> Set a higher explode value for the segment you want to pull out from the pie.</em>

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [3000, 6000, 2000, 4000]

explode = [0, 0.1, 0, 0]

plt.pie(
    spending,
    labels=categories,
    autopct="%1.1f%%",
    explode=explode
)

plt.title("Monthly Online Spending Distribution")

plt.show()
```

```python
# 3) Transform your pie chart into a donut chart to show the same spending data by adding a white circle at the center using 
# plt.Circle and ax.add_artist.

categories = ["Food Delivery", "Shopping", "Entertainment", "UPI Payments"]
spending = [3000, 6000, 2000, 4000]

explode = [0, 0.1, 0, 0]

fig, ax = plt.subplots()

ax.pie(
    spending,
    labels=categories,
    autopct="%1.1f%%",
    explode=explode
)
circle = plt.Circle((0, 0), 0.5, color="white")
ax.add_artist(circle)

plt.title("Monthly Online Spending Distribution")

plt.show()
```

```python
# 4) Suppose you want to compare the market share of 4 IPL cricket teams based on the number of Instagram followers. Build a donut chart
# where each segment shows the team name and follower percentage, and the chart title clearly states 'IPL Teams Instagram Followers 
# Share'.<br><br><em><strong>Constraint:</strong> Use both labels and autopct to display information on the chart.</em>

teams = ["RCB", "CSK", "MI", "KKR"]
followers = [24.8, 23, 20, 8.1]

fig, ax = plt.subplots()

ax.pie(
    followers,
    labels=teams,
    autopct="%1.1f%%"
)

circle = plt.Circle((0, 0), 0.5, color="white")
ax.add_artist(circle)

plt.title("IPL Teams Instagram Followers Share")

plt.show()
```
