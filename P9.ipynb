```python
# 1:
import pandas as pd

dates = pd.Series([
    "15-01-2026",
    "28-03-2026",
    "10-06-2025",
    "25-09-2024",
    "05-12-2026"
])

print(dates)
```

```python
dates = pd.to_datetime(dates, format="%d-%m-%Y")

print(dates)
```

```python
# 2:
df = pd.DataFrame({
    "movie": ["Dangal", "3 Idiots", "PK", "Pathaan"],
    "release_date": ["23-12-2016", "25-12-2009", "19-12-2014", "25-01-2023"]
})

print(df)
```

```python
df["release_date"] = pd.to_datetime(df["release_date"], format="%d-%m-%Y")

df["year"] = df["release_date"].dt.year
df["month"] = df["release_date"].dt.month
df["day"] = df["release_date"].dt.day

print(df)
```

```python
# 3:
dates = pd.date_range(start="today", periods=7)

print(dates)
```

```python
# 4:

df = pd.DataFrame({
    "user_id": [101, 102, 103, 104, 105],
    "start_date": [
        "01-01-2026",
        "15-02-2026",
        "10-04-2026",
        "20-06-2026",
        "01-08-2026"
    ]
})

print(df)
```

```python
df["start_date"] = pd.to_datetime(df["start_date"], format="%d-%m-%Y")

today = pd.Timestamp.today()

df["tenure_days"] = (today - df["start_date"]).dt.days

print(df)
```
