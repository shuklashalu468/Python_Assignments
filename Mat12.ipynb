```python
# 1) Download a sample sales dataset (for example, from Kaggle or use a small CSV you create) with columns like product, category, date, 
# and sales amount. Load it into a Pandas DataFrame and display the first 5 rows to verify the data.

import pandas as pd

data = {
    "product": ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse", "Monitor"],
    "category": ["Electronics", "Electronics", "Accessories", "Accessories", "Accessories", "Electronics"],
    "date": ["2026-01-01", "2026-01-02", "2026-01-03", "2026-01-04", "2026-01-05", "2026-01-06"],
    "sales_amount": [55000, 25000, 3000, 1500, 800, 12000]
}

df = pd.DataFrame(data)

df.to_csv("sales_data.csv", index=False)
```

```python
# 2) Using Matplotlib, create a bar chart that shows total sales for each product category (like 'Electronics', 'Clothing', 
# 'Groceries'), similar to how Flipkart or Myntra might track category performance.

import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

category_sales = df.groupby("category")["sales_amount"].sum()

plt.bar(category_sales.index, category_sales.values)

plt.xlabel("Product Category")
plt.ylabel("Total Sales (₹)")
plt.title("Total Sales by Product Category")

plt.show()
```

```python
# 3) Plot a line chart to visualize monthly sales trends over a year, using the date column to group sales by month. Add a title and 
# axis labels to make the chart presentation-ready.

df = pd.read_csv("sales_data.csv")

df["date"] = pd.to_datetime(df["date"])

monthly_sales = df.groupby(df["date"].dt.month)["sales_amount"].sum()

plt.figure(figsize=(10, 5))

plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.xlabel("Month")
plt.ylabel("Total Sales (₹)")
plt.title("Monthly Sales Trend")

plt.xticks(range(1, 13))
plt.grid(True)

plt.show()
```

```python
# 4) Create a histogram to show the distribution of order amounts (sales values), highlighting how many small vs. large purchases 
# happen — like analyzing order size patterns on Zomato or Swiggy.<br><br><em><strong>Hint:</strong> Use plt.hist() and set bins to
# 10 for a clear distribution.</em>

df = pd.read_csv("sales_data.csv")

plt.figure(figsize=(10, 5))

plt.hist(df["sales_amount"], bins=10)

plt.xlabel("Order Amount (₹)")
plt.ylabel("Number of Orders")
plt.title("Distribution of Order Amounts")

plt.show()
```

```python
# 5) Add value annotations to your category bar chart to display the exact sales amount above each bar, then export the chart as
# a PNG image file named 'category_sales_kpi.png'.

df = pd.read_csv("sales_data.csv")

category_sales = df.groupby("category")["sales_amount"].sum()

plt.figure(figsize=(10, 5))

plt.bar(category_sales.index, category_sales.values)

for i, value in enumerate(category_sales.values):
    plt.text(i, value + 500, f"₹{value}", ha="center")

plt.xlabel("Product Category")
plt.ylabel("Total Sales (₹)")
plt.title("Total Sales by Product Category")

plt.savefig("category_sales_kpi.png", dpi=150, bbox_inches="tight")

plt.show()
```
