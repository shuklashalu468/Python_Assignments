```python
# 1) Load a CSV file containing Flipkart product ratings and prices, then use Seaborn's jointplot with kind='scatter' to 
# visualize the relationship between rating and price.

import pandas as pd

df = pd.read_csv("flipkart_products.csv")

print(df.columns)

df["rating"] = [4.2, 4.5, 3.8, 4.7, 4.0, 4.3, 3.9, 4.6, 4.1, 4.4]

print(df.head())

import seaborn as sns
import matplotlib.pyplot as plt

sns.jointplot(
    data=df,
    x="rating",
    y="price",
    kind="scatter"
)

plt.show()
```
