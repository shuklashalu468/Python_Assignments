```python
import re

text = """
Contact me at 9876543210.
Price is 15000.
Another number is 8123456789.
Order ID: 1234567890.
Call 7654321098 for details.
"""

pattern = r'\b[789]\d{9}\b'

phone_numbers = re.findall(pattern, text)

print("Valid Indian phone numbers:")

for number in phone_numbers:
    print(number)
```

```python
import re

def check_date(text):
    pattern = r'\b\d{2}/\d{2}/\d{4}\b'
    
    if re.search(pattern, text):
        return True
    else:
        return False


text = input("Enter a string: ")

print(check_date(text))
```

```python
import re

text = """
Amazing food! Contact us at food@gmail.com
or zomato.user123@yahoo.com for more details.
For complaints, email support@restaurant.in.
"""

pattern = r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}'

emails = re.findall(pattern, text)

print(emails)
```

```python
import re

text = "My phone number is 9876543210 and my friend's number is 8765432109."

pattern = r'\b\d{6}(\d{4})\b'

masked_text = re.sub(pattern, r'******\1', text)

print(masked_text)
```

```python
import re

pattern = r'\bOD\d{18}\b'

order_strings = [
    "OD123456789012345000",
    "OD987654321098765432",
    "AB123456789012345678",
    "OD12345"
]

for order in order_strings:
    result = re.search(pattern, order)

    if result:
        print(order, "-> Valid Order ID")
    else:
        print(order, "-> Invalid Order ID")
```
