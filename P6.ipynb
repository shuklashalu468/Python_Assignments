```python
import pandas as pd
# 1:
df = pd.read_csv("spotify_streams.csv")

print("Original DataFrame:")
print(df)

df = df.drop_duplicates(subset=["song"], keep="first")

print("DataFrame after removing duplicates:")
print(df)
```

```python
# 2:
data = {
    "product_id": [101, 102, 101, 103, 102, 101, 104],
    "user_id": [1, 2, 1, 3, 2, 4, 5],
    "rating": [5, 4, 5, 3, 4, 2, 5],
    "review_text": [
        "Excellent product",
        "Good quality",
        "Excellent product",
        "Average product",
        "Good quality",
        "Not good",
        "Very nice product"
    ]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

duplicates_before = df.duplicated(
    subset=["product_id", "user_id"]
).sum()

print("Number of duplicate reviews before:")
print(duplicates_before)

df = df.drop_duplicates(
    subset=["product_id", "user_id"],
    keep="first"
)

duplicates_after = df.duplicated(
    subset=["product_id", "user_id"]
).sum()

print("Number of duplicate reviews after:")
print(duplicates_after)

print("DataFrame after removing duplicates:")
print(df)
```

```python
# 3:

data = {
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "order_amount": [250, 400, 350, 500, 450, 300, 5000, 380]
}

df = pd.DataFrame(data)

print("Zomato Orders:")
print(df)

Q1 = df["order_amount"].quantile(0.25)
Q3 = df["order_amount"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = df[
    (df["order_amount"] < lower_limit) |
    (df["order_amount"] > upper_limit)
]

print("Outlier Order IDs:")
print(outliers["order_id"])
```

```python
# 4:
from scipy.stats import zscore

data = {
    "date": ["2026-04-01", "2026-04-02", "2026-04-03", "2026-04-04", "2026-04-05", "2026-04-06", "2026-04-07"],
    "tickets_sold": [1200, 1350, 1100, 1400, 1250, 1300, 5000]
}

df = pd.DataFrame(data)

print("IPL Ticket Sales:")
print(df)

df["z_score"] = zscore(df["tickets_sold"])

print("Data with Z-score:")
print(df)

outliers = df[df["z_score"].abs() > 2]

print("Dates with Z-score greater than 2:")
print(outliers["date"])
```

```python
# 5:
data = {
    "transaction_id": [101, 102, 103, 104, 105, 106, 107],
    "transaction_amount": [500, 700, 450, 600, 550, 650, 5000]
}

df = pd.DataFrame(data)

print(df)
```

```python
def find_outliers(df):
    
    df["z_score"] = zscore(df["transaction_amount"])
    
    outliers = df[df["z_score"].abs() > 2]
    
    return outliers


result = find_outliers(df)

print("Outlier Transactions:")
print(result)
```
