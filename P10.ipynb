```python
# 1:
import pandas as pd

users = pd.DataFrame({
    "user_id": [101, 102, 103, 104, 105],
    "username": ["Rahul", "Priya", "Aman", "Neha", "Riya"],
    "city": ["Delhi", "Mumbai", "Pune", "Ahmedabad", "Jaipur"]
})

print(users)
```

```python
orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6, 7],
    "user_id": [101, 103, 101, 105, 102, 103, 102],
    "order_amount": [500, 800, 300, 1200, 450, 650, 900]
})

print(orders)
```

```python
result = pd.merge(users, orders, on="user_id", how="inner")

print(result)
```

```python
# 2:
result = pd.merge(users, orders, on="user_id", how="left")

print(result)
```

```python
# 3:

restaurants = pd.DataFrame({
    "restaurant_id": [101, 102, 103, 104, 105],
    "name": ["Dominos", "Burger King", "Biryani House", "Momos Point", "Pizza Hut"],
    "city": ["Delhi", "Mumbai", "Pune", "Ahmedabad", "Jaipur"]
})

print(restaurants)
```

```python
zomato_reviews = pd.DataFrame({
    "review_id": [1, 2, 3, 4, 5, 6],
    "restaurant_id": [101, 102, 101, 106, 103, 107],
    "rating": [5, 4, 4, 3, 5, 2],
    "reviewer_city": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Pune", "Chennai"]
})

print(zomato_reviews)
```

```python
result = pd.merge(
    restaurants,
    zomato_reviews,
    on="restaurant_id",
    how="outer"
)

print(result)
```

```python
# 4:
import pandas as pd

playlists = pd.DataFrame({
    "playlist_id": [1, 2, 3, 4, 5],
    "user_id": [101, 102, 103, 104, 105],
    "genre": ["Pop", "Rock", "Bollywood", "Hip-Hop", "Jazz"],
    "city": ["Delhi", "Mumbai", "Pune", "Ahmedabad", "Jaipur"]
})

user_profiles = pd.DataFrame({
    "user_id": [101, 102, 103, 104, 106],
    "username": ["Rahul", "Priya", "Aman", "Neha", "Riya"],
    "city": ["Delhi", "Mumbai", "Pune", "Delhi", "Jaipur"]
})
```

```python
result = pd.merge(
    playlists,
    user_profiles,
    on=["user_id", "city"],
    how="inner"
)

print(result)
```

```python
# 5:

products = pd.DataFrame({
    "product_id": [101, 102, 103, 104],
    "name": ["Mobile", "Laptop", "Headphones", "Watch"],
    "category": ["Electronics", "Electronics", "Electronics", "Accessories"]
})

ratings = pd.DataFrame({
    "rating_id": [1, 2, 3, 4, 5],
    "product_id": [101, 102, 105, 103, 105],
    "rating": [5, 4, 3, 5, 2]
})

print("Products:")
print(products)

print("Ratings:")
print(ratings)
```

```python
result = pd.merge(products, ratings, on="product_id")

print(result)
```

```python
result = pd.merge(
    products,
    ratings,
    on="product_id",
    how="right"
)

print(result)
```
