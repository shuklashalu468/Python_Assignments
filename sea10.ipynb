```python
# 1) Use sns.set() to apply the 'poster' context to a Seaborn barplot showing the number of songs streamed per day on Spotify, and 
# observe how the plot elements change compared to the default context.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "songs_streamed": [25, 32, 28, 40, 45, 55, 50]
}

df = pd.DataFrame(data)

sns.set(context="poster")

sns.barplot(
    data=df,
    x="day",
    y="songs_streamed"
)

plt.xlabel("Day")
plt.ylabel("Number of Songs Streamed")
plt.title("Daily Spotify Songs Streamed")

plt.show()
```

```python
# 2) Create a Seaborn countplot showing the distribution of food orders by cuisine type (e.g., Indian, Chinese, Italian) using a 
# 'colorblind' palette, similar to how Zomato might visualize order trends.<br><br><em><strong>Hint:</strong> 
# Use palette='colorblind' in your plot function.</em>

data = {
    "cuisine": [
        "Indian", "Chinese", "Italian", "Indian",
        "Chinese", "Indian", "Italian", "Chinese",
        "Indian", "Italian", "Chinese", "Indian"
    ]
}

df = pd.DataFrame(data)

sns.countplot(
    data=df,
    x="cuisine",
    palette="colorblind"
)

plt.xlabel("Cuisine Type")
plt.ylabel("Number of Orders")
plt.title("Food Orders by Cuisine Type")

plt.show()
```

```python
# 4) Refactor an existing Seaborn plot by switching its palette from 'deep' to 'colorblind' and changing the context from 'talk' to 
# 'poster'. Briefly describe the visual differences you notice after applying these changes. 

data = {
    "cuisine": [
        "Indian", "Chinese", "Italian", "Indian",
        "Chinese", "Indian", "Italian", "Chinese",
        "Indian", "Italian", "Chinese", "Indian"
    ]
}

df = pd.DataFrame(data)

sns.set(
    context="poster",
    palette="colorblind"
)

sns.countplot(
    data=df,
    x="cuisine",
    hue="cuisine",
    legend=False
)

plt.xlabel("Cuisine Type")
plt.ylabel("Number of Orders")
plt.title("Food Orders by Cuisine Type")

plt.show()
```

```python
# 5) Design a Seaborn theme that mimics the look of a Flipkart sales dashboard: use sns.set() to set a 'deep' palette, 'talk' context, 
# and customize the font scale and grid style. Apply your theme to a barplot showing sales of 5 product categories.

data = {
    "category": ["Mobiles", "Laptops", "Fashion", "Home Appliances", "Beauty"],
    "sales": [45000, 38000, 30000, 25000, 18000]
}

df = pd.DataFrame(data)

sns.set(
    context="talk",
    style="whitegrid",
    palette="deep",
    font_scale=1.1
)

sns.barplot(
    data=df,
    x="category",
    y="sales",
    hue="category",
    legend=False
)

plt.xlabel("Product Category")
plt.ylabel("Sales (₹)")
plt.title("Flipkart Sales by Product Category")

plt.show()
```
