```python
# 1) Use plt.bar() to create a vertical bar chart showing the number of daily orders for 5 popular food delivery apps (e.g., Zomato, 
# Swiggy, Uber Eats, Dunzo, Domino's) and set a custom color for each bar.

import matplotlib.pyplot as plt

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 150, 90, 70, 110]
colors = ["red", "orange", "blue", "green", "purple"]

plt.bar(apps, orders, color = colors)

plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Daily Orders")
plt.title("Daily Orders by Food Delivery App")

plt.show()
```

```python
#2) Create a horizontal bar chart using plt.barh() to display the number of followers for 6 different Instagram influencers; add a title
# and label the x-axis as 'Followers (in millions)'.

influencers = ["A", "B", "C", "D", "E", "F"]
followers = [12, 18, 25, 15, 30, 22]

plt.barh(influencers, followers)

plt.xlabel("Followers (in millions)")
plt.ylabel("Instagram Influencers")
plt.title("Instagram Influencers by Followers")

plt.show()
```

```python
# 3) Modify your vertical bar chart to display the exact order count as a text label above each bar using plt.text().<br><br><em>
# <strong>Hint:</strong> Use a for loop to iterate through the bars and place the text labels at the correct height.</em>

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 150, 90, 70, 110]
colors = ["red", "orange", "blue", "green", "purple"]

plt.bar(apps, orders, color=colors)

plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Daily Orders")
plt.title("Daily Orders by Food Delivery App")

for i in range(len(apps)):
    plt.text(i, orders[i] + 2, orders[i], ha="center")

plt.show()
```

```python
# 4) Change the width of the bars in your food delivery app chart to make them thinner, and adjust the spacing so the bars do not 
# overlap.<br><br><em><strong>Constraint:</strong> Use the 'width' parameter in plt.bar() and experiment with values less than 0.8.

apps = ["Zomato", "Swiggy", "Uber Eats", "Dunzo", "Domino's"]
orders = [120, 150, 90, 70, 110]
colors = ["red", "orange", "blue", "green", "purple"]

plt.bar(apps, orders, color=colors, width=0.5)

plt.xlabel("Food Delivery Apps")
plt.ylabel("Number of Daily Orders")
plt.title("Daily Orders by Food Delivery App")

plt.show()
```
