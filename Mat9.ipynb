```python
# 1) Create a stacked bar chart using Matplotlib that shows the number of food orders from Zomato and Swiggy for each day of the week. 
# Use different colors for each platform and add a legend.

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

zomato_orders = [120, 150, 130, 170, 180, 200, 190]
swiggy_orders = [110, 140, 125, 160, 175, 210, 185]

plt.bar(days, zomato_orders, color="red", label="Zomato")

plt.bar(days, swiggy_orders,
        bottom=zomato_orders,
        color="orange",
        label="Swiggy")

plt.xlabel("Days")
plt.ylabel("Number of Food Orders")
plt.title("Zomato and Swiggy Food Orders")

plt.legend()

plt.show()
```

```python
# 2) Plot an area chart that visualizes the daily active users on Spotify and YouTube over a 7-day period using Matplotlib's 
# fill_between(). Make sure each platform's area is filled with a different color and add labels for clarity.

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

spotify_users = [1200, 1350, 1280, 1500, 1600, 1800, 1700]
youtube_users = [2000, 2200, 2100, 2400, 2500, 2800, 2700]

plt.figure(figsize=(10, 5))

plt.fill_between(days, spotify_users, color="green", alpha=0.5, label="Spotify")
plt.fill_between(days, youtube_users, color="red", alpha=0.4, label="YouTube")

plt.xlabel("Days")
plt.ylabel("Daily Active Users")
plt.title("Spotify vs YouTube Daily Active Users")

plt.legend()

plt.show()
```

```python
# 3) Simulate the average delivery time (in minutes) for Flipkart orders over 10 days, and add error bars to represent the standard 
# deviation for each day's delivery time using Matplotlib's errorbar() function.<br><br><em><strong>Hint:</strong> Use the yerr 
# parameter to show error bars for each data point.</em>

days = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5",
        "Day 6", "Day 7", "Day 8", "Day 9", "Day 10"]

avg_delivery_time = [35, 38, 32, 40, 36, 42, 39, 34, 37, 41]

std_dev = [3, 4, 2, 5, 3, 4, 3, 2, 4, 5]

plt.figure(figsize=(10, 5))

plt.errorbar(
    days,
    avg_delivery_time,
    yerr=std_dev,
    fmt="o-",
    capsize=5
)

plt.xlabel("Days")
plt.ylabel("Average Delivery Time (Minutes)")
plt.title("Flipkart Average Delivery Time with Standard Deviation")

plt.show()
```

```python
# 4) Plot a time series line chart showing your daily Paytm wallet balance over the last 14 days using Matplotlib. Format the x-axis to 
# display dates in 'DD MMM' format (e.g., '12 Jun').

import matplotlib.dates as mdates
from datetime import datetime, timedelta

dates = [datetime.today() - timedelta(days=i) for i in range(13, -1, -1)]

balance = [2500, 2300, 2700, 2400, 3000, 2800, 3200,
           2900, 3100, 2750, 3400, 3300, 3600, 3500]

plt.figure(figsize=(10, 5))

plt.plot(dates, balance, marker="o")

plt.xlabel("Date")
plt.ylabel("Paytm Wallet Balance (₹)")
plt.title("Daily Paytm Wallet Balance - Last 14 Days")

plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%d %b"))

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
