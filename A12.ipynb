```python
# Lambda function to add 18% GST
add_gst = lambda price: price + (price * 0.18)

# Test the lambda function
print("Price after GST for Rs.100:", add_gst(100))
print("Price after GST for Rs.250:", add_gst(250))
print("Price after GST for Rs.500:", add_gst(500))
```

```python
# List of song titles
songs = [' shape OF you ', ' believer ', ' blinding LIGHTS ', ' excuses ', ' kesariya ']

# Use map() and lambda to clean the titles
clean_songs = list(map(lambda song: song.strip().title(), songs))

# Print the cleaned list
print("Cleaned Song Titles:")
print(clean_songs)
```

```python
# List of Flipkart product names
products = ["Samsung Galaxy", "Redmi Note", "sony Headphones", "Boat Speaker", "smart Watch"]

# Use filter() and lambda to select products starting with 'S' (case-insensitive)
filtered_products = list(filter(lambda product: product.lower().startswith("s"), products))

# Print the filtered products
print("Products starting with 'S':")
print(filtered_products)
```

```python
from functools import reduce

# List of order amounts
order_amounts = [120, 340, 560, 80]

# Use reduce() to calculate the total bill
total_bill = reduce(lambda x, y: x + y, order_amounts)

# Print the total bill
print("Total Bill Amount:", total_bill)
```

```python
from functools import reduce

# List of numbers
numbers = [40, 60, 80, 120]

# Step 1: Double each number using map()
doubled = list(map(lambda x: x * 2, numbers))

# Step 2: Keep only numbers greater than 100 using filter()
filtered = list(filter(lambda x: x > 100, doubled))

# Step 3: Sum the remaining numbers using reduce()
total = reduce(lambda x, y: x + y, filtered)

# Print the results
print("Original List:", numbers)
print("Doubled List:", doubled)
print("Filtered List (>100):", filtered)
print("Total Sum:", total)
```
